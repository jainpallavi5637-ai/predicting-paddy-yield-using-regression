```python
#Pearson correlation Test
from scipy.stats import pearsonr
import pandas as pd

features = [
    'Hectares ',
    'Seedrate(in Kg)',
    'Urea_40Days',
    'DAP_20days',
    'Trash(in bundles)'
]

results = []

for feature in features:
    corr, p_value = pearsonr(
        df[feature],
        df['Paddy yield(in Kg)']
    )

    inference = "Reject H₀" if p_value < 0.05 else "Fail to Reject H₀"

    results.append([
        feature,
        round(corr, 4),
        p_value,
        inference
    ])

results_df = pd.DataFrame(
    results,
    columns=['Feature', 'Correlation', 'P-value', 'Inference']
)

print(results_df)
```

```python
#Linear Regression
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
```

```python
# Accuracy Evaluation
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import numpy as np

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print("Linear Regression Results")
print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R2 Score:", r2)
```

```python
from sklearn.metrics import r2_score, mean_absolute_error, root_mean_squared_error
import numpy as np

models = {
    'Linear Regression': model,
    'Random Forest': rf_model,
    'XGBoost': xgb_model,
    'Decision Tree': dt_model

}

results = []

for name, model_instance in models.items():

    from sklearn.base import clone
    new_model = clone(model_instance)
    new_model.fit(X_train_new, y_train_new)

    preds_new = new_model.predict(X_test_new)
    r2_new = r2_score(y_test_new, preds_new)
    mae_new = mean_absolute_error(y_test_new, preds_new)
    rmse_new = root_mean_squared_error(y_test_new, preds_new)

    results.append({
        'Model': name,
        'Original R² (With Hectares)': 0.90,
        'New R² (Per Hectare)': round(r2_new, 3),
        'New MAE': round(mae_new, 3),
        'New RMSE': round(rmse_new, 3)
    })

importances = new_model.feature_importances_
feature_names = X_train_new.columns

df_importance = pd.DataFrame({
    'Feature': feature_names,
    'Importance': importances
}).sort_values(by='Importance', ascending=False)

print("\n--- Actual Feature Importance (Efficiency Model) ---")
print(df_importance)
```
